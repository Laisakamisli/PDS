### Tugas visualisasi dataset 97 kota

```python
import pandas as pd

data2 = pd.read_csv('https://raw.githubusercontent.com/Alam-A99/DATAAKSI/refs/heads/main/dataset_fiskal_97_hipoFull.csv')
data2.head()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Membaca dataset
url = 'https://raw.githubusercontent.com/Alam-A99/DATAAKSI/refs/heads/main/dataset_fiskal_97_hipoFull.csv'
data2 = pd.read_csv(url)

# Filter data khusus Kota Banda Aceh
# (Sesuaikan nama kolom wilayah/daerah jika berbeda pada dataset, misalnya 'Kabupaten/Kota' atau 'Nama Wilayah')
data_aceh = data2[data2['Kota'] == 'Kota Banda Aceh'].copy()

# Urutkan berdasarkan tahun
data_aceh = data_aceh.sort_values(by='Tahun')

# Pengaturan Visualisasi
plt.style.use('Solarize_Light2')
plt.figure(figsize=(9, 5))

# Buat Barplot
sns.barplot(
    data=data_aceh,
    x='Tahun',
    y='KEMISKINAN',
    palette='PuRd',  # Palet ungu-pink pastel
    hue='Tahun',
    legend=False,
)

# Buat Line Plot (Garis Tren)
plt.plot(
    range(len(data_aceh)),
    data_aceh['KEMISKINAN'],
    color='#FFBDE3',  # Warna pastel pink lembut
    marker='o',
    markersize=8,
    linestyle='-',
    linewidth=2.5,
    label='Tren',
)

# Menambahkan angka di atas batang
for i, val in enumerate(data_aceh['KEMISKINAN']):
    plt.text(
        i,
        val + (val * 0.01),
        f'{val:.2f}',
        ha='center',
        va='bottom',
        fontsize=9,
        fontweight='bold',
        color='#4A4A4A',
    )

# Judul dan Label Sumbu
plt.title(
    'Tren Kemiskinan di Kota Banda Aceh per Tahun',
    fontsize=14,
    fontweight='bold',
    pad=15,
)
plt.xlabel('Tahun', fontsize=11)
plt.ylabel('Tingkat / Jumlah Kemiskinan', fontsize=11)
plt.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

```python

import numpy as np
import pandas as pd
import statsmodels.api as sm
#
data = {
    "partisipan": np.arange(1,11),
    "cookies": [50,40,60,55,45,65,70,75,80,90],
    "tepung": [20,25,30,35,40,50,55,60,65,70],
    "air": [60,55,65,70,62,75,80,85,90,95]
}

df = pd.DataFrame(data)

x = df[['cookies','tepung']]
y = df['air']

x = sm.add_constant(x)

model = sm.OLS(y,x)
result = model.fit()

print(result.summary())
```
